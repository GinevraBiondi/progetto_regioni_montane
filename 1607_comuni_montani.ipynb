# Discussione dei risultati

## Una tipologia dei territori oltre lo spopolamento

L'analisi di clustering evidenzia come i comuni considerati non costituiscano un insieme omogeneo, ma possano essere raggruppati in differenti profili territoriali caratterizzati da specifiche combinazioni di fattori demografici, geografici, climatici ed economici. Questo risultato suggerisce che il fenomeno dello spopolamento non possa essere interpretato esclusivamente come conseguenza della montanità o della distanza dai principali centri urbani, ma debba essere analizzato come il risultato dell'interazione tra molteplici dimensioni territoriali.

L'approccio adottato consente quindi di superare una lettura dicotomica tra aree in crescita e aree in declino, evidenziando come esistano differenti percorsi evolutivi che riflettono diversi livelli di vulnerabilità e resilienza.

## I profili territoriali individuati

L'analisi ha identificato otto cluster, ciascuno caratterizzato da una specifica configurazione delle variabili considerate.

### Cluster 0 – Comuni montani in moderata crescita

Il primo cluster comprende prevalentemente comuni montani o collinari che mostrano una moderata crescita demografica, una discreta distanza dai principali poli urbani e livelli intermedi di rischio idrogeologico. Il mercato immobiliare rimane relativamente contenuto, mentre gli effetti del cambiamento climatico risultano presenti ma meno marcati rispetto ad altri gruppi. Questi territori rappresentano realtà che, pur mantenendo caratteristiche tipiche della montagna, conservano una buona capacità di attrazione.

### Cluster 1 – Comuni montani maturi

Il secondo cluster raggruppa comuni situati alle quote più elevate, caratterizzati da una lieve diminuzione della popolazione e da una significativa riduzione delle precipitazioni nevose. Il cambiamento climatico emerge come elemento distintivo di questo gruppo, evidenziando la crescente vulnerabilità delle aree alpine, pur in presenza di condizioni socioeconomiche relativamente stabili.

### Cluster 2 – Comuni urbani o turistici ad alta valorizzazione immobiliare

Questo cluster comprende comuni situati prevalentemente nelle aree di pianura o nei principali poli turistici, caratterizzati dai valori immobiliari più elevati dell'intero campione. Nonostante una leggera contrazione della popolazione residente, tali territori mantengono un'elevata attrattività economica, confermata dal valore del mercato immobiliare e dalla prossimità ai principali centri urbani.

### Cluster 3 – Comuni montani dinamici ma climaticamente vulnerabili

I comuni appartenenti a questo gruppo mostrano una crescita demografica positiva, ma contemporaneamente registrano gli incrementi più elevati della temperatura media e una consistente riduzione delle precipitazioni nevose. Il cluster rappresenta quindi territori che mantengono una buona capacità di sviluppo pur essendo particolarmente esposti agli effetti del cambiamento climatico.

### Cluster 4 – Comuni fragili

Il quarto cluster identifica le situazioni di maggiore criticità. I comuni appartenenti a questo gruppo registrano il più elevato calo demografico, una forte esposizione al rischio idrogeologico e valori immobiliari relativamente bassi. L'associazione tra vulnerabilità ambientale, declino demografico e ridotta attrattività economica suggerisce la presenza di processi di fragilizzazione territoriale che potrebbero richiedere politiche di intervento specifiche.

### Cluster 5 – Grandi poli urbani

Questo cluster comprende i principali centri urbani del campione, caratterizzati da elevata accessibilità, ampia popolazione di riferimento e un mercato immobiliare più sviluppato rispetto alla media. La loro presenza costituisce un'importante categoria di confronto rispetto ai comuni montani e periferici.

### Cluster 6 – Comuni periurbani in espansione

I comuni appartenenti a questo gruppo si collocano prevalentemente nelle aree periurbane e mostrano una crescita demografica positiva, favorita dalla vicinanza ai principali poli urbani. Allo stesso tempo, tali territori registrano il maggiore incremento delle precipitazioni, evidenziando come anche le aree maggiormente dinamiche siano interessate dagli effetti del cambiamento climatico.

### Cluster 7 – Comuni suburbani ad alta attrattività

L'ultimo cluster raccoglie comuni caratterizzati dalla maggiore crescita demografica osservata, da un'elevata accessibilità ai poli urbani e da un rischio idrogeologico relativamente contenuto. Questi territori rappresentano le realtà più attrattive del campione, nelle quali la vicinanza ai servizi e alle opportunità economiche favorisce il mantenimento o l'incremento della popolazione residente.

# Una lettura in termini di vulnerabilità e resilienza

L'interpretazione congiunta dei cluster consente di individuare tre macro-profili territoriali.

Il primo comprende i **territori resilienti**, rappresentati principalmente dai cluster 0, 3, 6 e 7. Questi comuni mostrano dinamiche demografiche positive, una buona accessibilità e una capacità di adattamento che consente loro di mantenere attrattività nonostante le trasformazioni climatiche in corso.

Il secondo è costituito dai **territori vulnerabili**, rappresentati soprattutto dal cluster 4. In questi comuni si concentrano simultaneamente spopolamento, elevato rischio idrogeologico e ridotta valorizzazione immobiliare, delineando un quadro di fragilità multidimensionale.

Infine, i cluster 1, 2 e 5 rappresentano **territori consolidati**, caratterizzati da condizioni relativamente stabili ma con differenti traiettorie evolutive. In particolare, il cluster 1 evidenzia una crescente esposizione agli effetti del cambiamento climatico, mentre i cluster 2 e 5 riflettono il ruolo dei principali poli urbani e turistici nella strutturazione del territorio.

# Implicazioni per le politiche territoriali

I risultati suggeriscono che le politiche rivolte ai territori montani dovrebbero adottare un approccio differenziato, evitando interventi uniformi per realtà profondamente eterogenee. Le strategie di sviluppo dovrebbero essere calibrate sulla base delle specifiche caratteristiche dei diversi profili territoriali.

Nei territori resilienti, gli interventi potrebbero concentrarsi sul consolidamento delle dinamiche positive, attraverso il potenziamento dei servizi, della mobilità e delle opportunità economiche. Nei territori vulnerabili risulta invece prioritario intervenire sulla riduzione del rischio idrogeologico, sul contrasto allo spopolamento e sul rafforzamento dell'accessibilità ai servizi essenziali. Per i territori consolidati, infine, assume particolare rilevanza la gestione degli effetti del cambiamento climatico e la valorizzazione sostenibile delle risorse locali.

# Conclusioni

L'analisi dimostra come la montagna italiana non possa essere interpretata come un sistema territoriale omogeneo. I processi di crescita, declino e trasformazione dipendono dall'interazione tra caratteristiche geografiche, accessibilità, cambiamento climatico e dinamiche socioeconomiche.

L'utilizzo di tecniche di clustering non supervisionato consente di individuare differenti profili territoriali che descrivono livelli diversi di vulnerabilità e resilienza. Questa classificazione rappresenta uno strumento utile sia per comprendere le dinamiche territoriali sia per supportare la progettazione di politiche pubbliche maggiormente mirate, orientate alle specifiche esigenze dei diversi contesti locali.

```python
meteo_year = (
    meteo
    .groupby(
        ["comune", "year"],
        as_index=False
    )
    .agg(
        temp_mean=("temperature_mean", "mean"),
        rain_sum=("rain_sum", "sum"),
        snow_sum=("snowfall_sum", "sum")
    )
)
##std tempreratura e pioggia
meteo_var = (
    meteo
    .groupby(
        ["comune", "year"],
        as_index=False
    )
    .agg(
        temp_sd=("temperature_mean", "std"),
        rain_sd=("rain_sum", "std")
    )
)
```

```python
meteo_year = meteo_year.merge(
    meteo_var,
    on=["comune", "year"]
)
##numero mesi con neve
snow_months = (
    meteo
    .assign(
        snow_month=lambda x:
        (x["snowfall_sum"] > 0).astype(int)
    )
    .groupby(
        ["comune", "year"],
        as_index=False
    )
    .agg(
        n_snow_months=("snow_month", "sum")
    )
)
meteo_year = meteo_year.merge(
    snow_months,
    on=["comune", "year"]
)
```

```python
##temp media invernale
winter = meteo[
    meteo["month"].isin([12, 1, 2])
]

winter_temp = (
    winter
    .groupby(
        ["comune", "year"],
        as_index=False
    )
    .agg(
        winter_temp_mean=("temperature_mean", "mean")
    )
)
#pioggia estrema: mesi sopra il 90 percentile
threshold = meteo["rain_sum"].quantile(0.90)

meteo["heavy_rain"] = (
        meteo["rain_sum"] > threshold
).astype(int)
```

```python
heavy_rain = (
    meteo
    .groupby(
        ["comune", "year"],
        as_index=False
    )
    .agg(
        n_heavy_rain_months=("heavy_rain", "sum")
    )
)
meteo_year = (
    meteo_year
    .merge(
        winter_temp,
        on=["comune", "year"],
        how="left"
    )
    .merge(
        heavy_rain,
        on=["comune", "year"],
        how="left"
    )
)
```

```python
meteo_year = (
    meteo_year
    .sort_values(
        ["comune", "year"]
    )
)
##aggiungo la media rolling degli ultimi 3 anni
meteo_year["snow_3y"] = (
    meteo_year
    .groupby("comune")
    ["snow_sum"]
    .transform(
        lambda x:
        x.rolling(
            3,
            min_periods=1
        ).mean()
    )
)
```

```python
##trend neve, temp, pioggia con regressione lineare: prendo la slope di y = beta*x + alfa
import numpy as np
import pandas as pd


def compute_trend(group, var):
    x = group["year"].values
    y = group[var].values

    if len(group) < 2:
        return np.nan

    # regressione lineare
    beta = np.polyfit(x, y, 1)[0]
    return beta


snow_trend = (
    meteo_year
    .groupby("comune")
    .apply(lambda g: compute_trend(g, "snow_sum"))
    .reset_index()
)

snow_trend.columns = ["comune", "snow_trend"]
temp_trend = (
    meteo_year
    .groupby("comune")
    .apply(lambda g: compute_trend(g, "temp_mean"))
    .reset_index()
)

temp_trend.columns = ["comune", "temp_trend"]
rain_trend = (
    meteo_year
    .groupby("comune")
    .apply(lambda g: compute_trend(g, "rain_sum"))
    .reset_index()
)

rain_trend.columns = ["comune", "rain_trend"]
meteo_trends = (
    snow_trend
    .merge(temp_trend, on="comune")
    .merge(rain_trend, on="comune")
)

meteo_year = meteo_year.rename(columns={"year": "anno"})
#uso solo il periodo 2014-2026
start_year = 1980
end_year = 2026


meteo_year = meteo_year[(meteo_year["anno"] >= start_year) & (meteo_year["anno"] <= end_year)]
demo = demo[(demo["anno"] >= start_year) & (demo["anno"] <= end_year)]
```

```python
from sklearn.metrics import silhouette_score, davies_bouldin_score
from scipy.stats import pearsonr, spearmanr

def get_sse(X, labels):
    sse = 0
    for l in np.unique(labels):
        c = np.mean(X[labels == l], axis=0)
        sse += np.sum(cdist(np.expand_dims(c, axis=0), X[labels == l]))
    return sse

def get_bss(X, labels):
    cluster_centers = list()
    for l in np.unique(labels):
        c = np.mean(X[labels == l], axis=0)
        cluster_centers.append(c)
    cluster_centers = np.array(cluster_centers)
    _, sizes = np.unique(labels, return_counts=True)
    m = np.mean(X, axis=0)
    bss = np.sum(sizes * cdist(np.expand_dims(m, axis=0), cluster_centers))
    return bss

def get_ideal_matrix(X, labels):
    P = np.ones((len(X), len(X)))
    for i in np.arange(len(X)):
        for j in np.arange(len(X)):
            if labels[i] == labels[j]:
                P[i, j] = 0.
    return P

def eval_clustering(X, D, labels):
   # corr = spearmanr(D.ravel(), get_ideal_matrix(X, labels).ravel())[0]
    sse = get_sse(X[labels != -1], labels[labels != -1])
    bss = get_bss(X[labels != -1], labels[labels != -1])
    silh = silhouette_score(X[labels != -1], labels[labels != -1])
    davis = davies_bouldin_score(X, labels)
    return (#corr,
            sse, bss, silh, davis)
```

Parallelamente, la quota di giovani (`share_giovani`) emerge come secondo fattore più importante, evidenziando il ruolo della componente giovane come indicatore di vitalità e capacità di rigenerazione demografica.

# 2. L'accessibilità conta, ma non è il fattore dominante

La variabile relativa al tempo necessario per raggiungere un hub urbano (`tempo_hub_5000`) presenta un'importanza significativa, ma inferiore rispetto alle variabili demografiche.

Questo risultato suggerisce che la marginalità geografica non sia sufficiente, da sola, a spiegare il declino demografico.

La distanza dai servizi e dai centri urbani rappresenta quindi un fattore di pressione, ma il suo effetto dipende dalla capacità del territorio di mantenere una struttura sociale ed economica attiva.

# 3. Il cambiamento climatico rappresenta un fattore secondario ma rilevante

Le variabili climatiche:

- `delta_snow`;
- `delta_temp`;
- `delta_rain`;

presentano complessivamente un'importanza inferiore rispetto alle variabili demografiche.

Questo suggerisce che il cambiamento climatico non agisca necessariamente come causa primaria dello spopolamento, ma piuttosto come un elemento che può amplificare fragilità già presenti.

Ad esempio, la riduzione della neve può avere effetti particolarmente rilevanti nei comuni fortemente dipendenti dal turismo invernale, mentre territori economicamente più diversificati possono risultare maggiormente capaci di adattarsi.

# 4. Relazione con i cluster individuati

I risultati della Random Forest confermano e approfondiscono l'interpretazione ottenuta tramite il clustering.

## Cluster 4 – Comuni fragili

Il cluster caratterizzato da maggiore vulnerabilità territoriale può essere interpretato come il risultato della combinazione di:

- elevata quota di popolazione anziana;
- ridotta presenza giovanile;
- bassa densità abitativa;
- minore attrattività economica.

## Cluster 7 – Comuni suburbani attrattivi

I comuni caratterizzati da crescita demografica positiva sono invece associati a:

- maggiore presenza di giovani;
- maggiore densità abitativa;
- migliore accessibilità;
- maggiore capacità di attrazione residenziale.

## Cluster montani

I cluster montani mostrano che la quota e il cambiamento climatico non determinano automaticamente il declino demografico. Alcuni territori montani mantengono dinamiche positive grazie alla presenza di condizioni socioeconomiche favorevoli e alla capacità di attrarre nuovi residenti.

# Conclusione

La Random Forest evidenzia che le dinamiche demografiche dei comuni analizzati sono guidate principalmente dalla struttura sociale interna dei territori, mentre gli aspetti climatici e ambientali assumono un ruolo complementare.

Il risultato principale è che lo spopolamento non appare come una conseguenza diretta della sola marginalità geografica o del cambiamento climatico, ma come il risultato dell'interazione tra:

- struttura per età della popolazione;
- capacità di attrarre giovani residenti;
- accessibilità ai servizi;
- condizioni economiche locali;
- pressioni ambientali.

Questa evidenza rafforza l'interpretazione dei cluster come differenti traiettorie territoriali, caratterizzate da diversi livelli di vulnerabilità e resilienza.
