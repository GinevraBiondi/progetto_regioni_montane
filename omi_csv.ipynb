```python
from pathlib import Path
import pandas as pd

# Cartella contenente i file
folder = Path("OMI_SE/2004-2025/dati_omi_grezzi")

dfs = []

for file in folder.glob("QI_*_VALORI.csv"):
    print(f"Leggo {file.name}")

    df = pd.read_csv(
        file,
        sep=";",
        encoding="latin1",
        dtype=str
    )

    # tiene traccia del file di origine
    df["file_origine"] = file.name

    dfs.append(df)

# Unisco tutti i dataframe
quotazioni = pd.concat(dfs, ignore_index=True)

print(f"Numero righe: {len(quotazioni):,}")
#print(f"Numero comuni: {quotazioni['Comune_descrizione'].nunique()}")

# Salvo il dataset aggregato
quotazioni.to_csv(
    "OMI.csv",
    sep=";",
    index=False,
    encoding="utf-8-sig"
)

print("CSV aggregato salvato.")
```
